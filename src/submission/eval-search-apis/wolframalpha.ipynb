````python
import nest_asyncio
nest_asyncio.apply()

from xag329a_a3.wolfram_api import Client

client = Client(app_id)
math_query = "integral cos(x)/sqrt(x) from 0 to 1"
query = 'What is the capital of France?'
result = client.query(math_query)
print(result)
````
