This notebook is for testing [tavily](https://www.tavily.com/) search API and API key setup.

Sign up at [tavily](https://www.tavily.com/), and the "Researcher" plan is **free** with limited usage as of 2/23/2026.

Once you sign up and log in, you can find your API keys from this [link](https://app.tavily.com/home).

Copy and add that to your `.env` file like this:

```bash
TAVILY_API_KEY=YOUR_API_KEY
```

Python API documentation link is [here](https://docs.tavily.com/sdk/python/reference).

````python
import os
from IPython.display import HTML
import textwrap
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()
````

````python
api_key = os.getenv("TAVILY_API_KEY") # env var is set in .env file
````

````python
from tavily import TavilyClient

tavily_client = TavilyClient(api_key=api_key)
response = tavily_client.search("Who is Leo Messi?")

print(response)
````

````python
len(response)
````
