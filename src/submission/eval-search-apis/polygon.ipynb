````python
import os
from IPython.display import HTML
import textwrap
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()
````

````python
api_key = os.getenv("POLYGON_API_KEY") # env var is set in .env file
````

````python
from massive import RESTClient
````

````python
client = RESTClient(api_key=api_key)
````

````python
ticker = "AAPL"
date = "2026-02-20"

try:
    result = client.get_daily_open_close_agg(ticker=ticker, date=date)
    print ({
        "date": result.from_, "open": result.open, "high": result.high,
        "low": result.low, "close": result.close, "volume": result.volume,
    })
except Exception as e:
    print( {"error": f"Error fetching stock data for {ticker}: {e}"})
````
